```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
# Input
X = np.array([[1],
              [2],
              [3],
              [4],
              [5]])

# Output
y = np.array([30, 40, 50, 60, 70])
```

```python
# Create model
model = LinearRegression()
```

```python
# Train model
model.fit(X, y)
```

```python
# Check coefficient and intercept
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Prediction
prediction = model.predict([[6]])

print("Predicted Score:", prediction[0])
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Features
X = np.array([
    [1, 1, 22],
    [2, 1, 25],
    [3, 2, 28],
    [5, 2, 32],
    [7, 3, 35]
])

# Target
y = np.array([30000, 35000, 45000, 60000, 75000])
```

```python
# Create model
model = LinearRegression()
```

```python
# Train model
model.fit(X, y)
```

```python
# Coefficients
print("Coefficients:", model.coef_)

# Intercept
print("Intercept:", model.intercept_)
```

```python
# Prediction
new_person = [[4, 2, 30]]

prediction = model.predict(new_person)

print("Predicted Salary:", prediction[0])
```
